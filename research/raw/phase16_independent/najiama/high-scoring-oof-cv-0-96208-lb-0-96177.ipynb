```python
import pandas as pd
from sklearn.metrics import roc_auc_score
import matplotlib.pyplot as plt

BLEND = '04'
TARGET = 'satisfaction'

# Load Train Data and Target
train_df = pd.read_csv("/kaggle/input/competitions/playground-series-s6e10/train.csv")
train_df[TARGET] = train_df[TARGET].astype(int)
y = train_df[TARGET].values

oof_data = pd.read_csv(f'/kaggle/input/s6e10-oof/{BLEND}_blend_oof.csv')
submission = pd.read_csv(f'/kaggle/input/s6e10-oof/{BLEND}_submission.csv')

ensemble_auc = roc_auc_score(y, oof_data[TARGET])
```

```python
print(f"✅ OOF Train Shapes: {oof_data.shape}")
print(f"✅ Test Submission Shapes: {submission.shape}\n")
print(f"🔥 Validated OOF ROC-AUC Score: {ensemble_auc:.5f}\n")
```
